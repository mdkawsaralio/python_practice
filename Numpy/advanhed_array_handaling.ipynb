```python
#handling missing values in numpy array using np.nan and np.isnan() method
```

```python
customer=np.array([1001,1002,1003,1004,1005,np.nan,1007,1008,np.nan,1010]) #nan is used to represent missing values in numpy array
missing_values=np.isnan(customer) #check for missing values in the array
missing_values
```

```python
print("customer total missing values: ",np.sum(missing_values)) #print the missing values in the array
```

```python
#handling missing values using np.nan_to_num() method

filled_with_zero=np.nan_to_num(customer) #replace missing values with 0
filled_with_zero
```

```python
filled_with_mean=np.nanmean(customer) #replace missing values with mean of the array
filled_with_mean   #not the expected output because mean is calculated without considering the missing values and it will not change the original data
```

```python
filled_with_mean=np.where(np.isnan(customer),np.nanmean(customer),customer) #replace missing values with mean of the array where true =nanmean(customer) and false = customer
filled_with_mean
```

```python
#if we want to remove the missing values from the array we can use np.isnan() method to check for missing values and then use boolean indexing to remove the missing values from the array.
cleaned_array=customer[~np.isnan(customer)] #remove missing values from the array
cleaned_array
```

```python
customer #not changed the original data because we have created a new array with cleaned data and it will not change the original data.
```

```python
#save the numpy array into a file using np.save() method and load the numpy array from a file using np.load() method

np.save('customer.npy', customer)
```

```python
np.load('customer.npy') #load the numpy array from a file
```

```python
#save numpy array into a text file using np.savetxt() method and load the numpy array from a text file using np.loadtxt() method

np.savetxt('customer.txt', customer) #save the numpy array into a text file
```

```python
np.loadtxt('customer.txt') #load the numpy array from a text file
```
