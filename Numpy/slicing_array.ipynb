```python
monthly_revenue = np.array([100, 200, 300,400,500,600,700,800,900,1000,1100,1200])
```

```python
monthly_revenue [1:5] #beginning from 1st index to 5th index
```

```python
monthly_revenue [5::] #beginning from 5th index to end
```

```python
monthly_revenue [5:-2] #beginning from 5th index to 2nd last index
```

```python
avg=np.mean(monthly_revenue)   #average of monthly revenue
avg
```

```python
monthly_above_avg = monthly_revenue [monthly_revenue > avg] #monthly revenue greater than average
monthly_above_avg
```

```python
import numpy as np

def calculate_nps(score):
    if score >= 9:
        return "Promoter"
    elif score >= 7:
        return "Passive"
    else:
        return "Detractor"

nps_ufunc = np.vectorize(calculate_nps)
print(nps_ufunc([10, 8, 6, 9, 7, 5]))
```
